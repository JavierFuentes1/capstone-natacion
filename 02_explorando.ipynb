## Paso 1 — Cargar

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.width", 150)
pd.set_option("display.max_columns", 30)

df = pd.read_csv("datos/nados.csv")

print("Filas:   ", f"{len(df):,}")
print("Columnas:", df.shape[1])
```

`df` es un **DataFrame**: una tabla, como una hoja de Excel pero manejada con código.
La `f` de `df` viene de *data frame*; es el nombre que usa todo el mundo por costumbre.

`df.shape` te da (filas, columnas). `df.shape[1]` es solo las columnas.

## Paso 4 — Filtrar filas

Esto es lo más útil de pandas. Escribes una **condición** y pandas te deja solo las
filas que la cumplen.

```python
# Solo hombres
hombres = df[df["sexo"] == "M"]
print(f"{len(hombres):,} nados de hombres, de {len(df):,} totales")
```

```python
# Miremos qué pasa por dentro: la condición sola es una columna de True/False
(df["sexo"] == "M").head(5)
```

Eso se llama **máscara booleana**. `df[mascara]` conserva las filas donde dice `True`.

Para combinar condiciones se usa `&` (y) y `|` (o), **con cada condición entre
paréntesis**. Es el error más común al empezar: si olvidas los paréntesis, Python
se queja.

```python
# 50 libre, hombres, piscina de 50 m
libre50 = df[(df["sexo"] == "M") &
             (df["distancia"] == 50) &
             (df["estilo"] == "FREESTYLE") &
             (df["piscina"] == "LCM")]

print(f"{len(libre50):,} nadadores")
libre50[["rank", "nombre", "pais", "tiempo", "edad"]].head(10)
```

## Paso 5 — Ordenar

`sort_values` ordena por una columna. `ascending=True` es de menor a mayor.
En natación, menor tiempo = más rápido.

```python
# Los 10 más rápidos del mundo en 50 libre 2024
(libre50
 .sort_values("tiempo_s")
 .head(10)[["nombre", "pais", "tiempo", "edad", "competencia"]])
```

## Paso 6 — Contar

`value_counts()` cuenta cuántas veces aparece cada valor. Es de lo que más se usa
en exploración.

```python
# Los 15 países con más nadadores en la muestra
df["pais"].value_counts().head(15)
```

```python
# Cuántos nados hay de cada prueba
df["prueba"].value_counts().head(10)
```

## Paso 7 — Buscar a Chile

`.str.contains()` busca texto dentro de una columna de texto.
Aquí usamos el código de país, que es exacto.

```python
chile = df[df["pais_codigo"] == "CHI"]
print(f"Nados chilenos: {len(chile):,}")
print(f"Nadadores chilenos distintos: {chile['person_id'].nunique()}")
print()
print(chile["prueba"].value_counts().head(10))
```

```python
# Los 15 mejores tiempos chilenos de la muestra, por puntos FINA
# (los puntos FINA permiten comparar pruebas distintas entre sí)
(chile
 .sort_values("puntos_fina", ascending=False)
 .head(15)[["nombre", "prueba", "piscina", "tiempo", "puntos_fina", "edad"]])
```

## Paso 8 — El primer gráfico

Un **histograma** muestra cómo se reparten los valores: cuántos nadadores caen en
cada rango de tiempo.

```python
fig, ax = plt.subplots(figsize=(9, 4.5))

ax.hist(libre50["tiempo_s"], bins=200, color="#0B6E75", edgecolor="white", linewidth=.4)

ax.set_xlabel("Tiempo en segundos")
ax.set_ylabel("Cantidad de nadadores")
ax.set_title("50 m libre masculino 2024 — cómo se reparten los tiempos",
             fontsize=12, loc="left")
ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="y", alpha=.25)
plt.tight_layout()
plt.show()
```

Mira la forma. No es una campana simétrica: hay una cola larga hacia la derecha.
Poquísimos nadadores muy rápidos a la izquierda, y muchos repartidos hacia tiempos
más lentos.

Esa asimetría tiene un nombre —**distribución sesgada a la derecha**— y es la razón
por la que en el análisis usamos **mediana** y no promedio: el promedio se deja
arrastrar por esa cola.

Prueba a cambiar `bins=60` por `bins=15` o `bins=200` y mira cómo cambia la lectura.

## Paso 9 — Comparar dos grupos

Dos histogramas superpuestos, con transparencia para poder ver los dos.

```python
h = df[(df["distancia"] == 100) & (df["estilo"] == "FREESTYLE") &
       (df["piscina"] == "LCM") & (df["sexo"] == "M")]
m = df[(df["distancia"] == 100) & (df["estilo"] == "FREESTYLE") &
       (df["piscina"] == "LCM") & (df["sexo"] == "F")]

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.hist(h["tiempo_s"], bins=60, alpha=.65, label=f"Hombres (n={len(h):,})", color="#0072B2")
ax.hist(m["tiempo_s"], bins=60, alpha=.65, label=f"Mujeres (n={len(m):,})", color="#D55E00")

ax.set_xlabel("Tiempo en segundos")
ax.set_ylabel("Cantidad de nadadores")
ax.set_title("100 m libre, piscina de 50 m, 2024", fontsize=12, loc="left")
ax.legend(frameon=False)
ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="y", alpha=.25)
plt.tight_layout()
plt.show()
```

### Ojo con el borde derecho

Fíjate en cómo las dos distribuciones se **cortan de golpe** por la derecha, como si
alguien hubiera tajado el gráfico. Eso no es real: es nuestra descarga.

Bajamos los 5.000 nadadores más rápidos de cada prueba. Los que están más allá de
ese corte existen —hay 19.812 hombres en el 100 libre— pero nosotros no los tenemos.

Es un buen recordatorio: **un gráfico también muestra cómo obtuviste los datos**, no
solo lo que pasa en el mundo. Esa pared de la derecha es la huella de nuestra decisión
de muestreo, y es la razón por la que en el informe declaramos que los resultados
describen a la élite.

## Paso 10 — Resumir por grupos

`groupby` parte la tabla en montoncitos y calcula algo de cada uno.

```python
# Mediana de edad por prueba y sexo
(df.groupby(["sexo", "distancia", "estilo"])["edad"]
   .median()
   .unstack(["distancia"])
   .round(1))
```

```python
# Cuántos nadadores aporta cada país, y su mejor puntaje FINA
(df.groupby("pais")
   .agg(nadadores=("person_id", "nunique"),
        mejor_fina=("puntos_fina", "max"))
   .sort_values("nadadores", ascending=False)
   .head(15))
```
